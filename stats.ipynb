```python
import pandas as pd
```

## Available models and domains

```python
df = pd.read_csv("https://thredds.climate.ifca.es/thredds/fileServer/inventories/cava.csv")
subset = df[df["location"].str.contains("fao/interp")][["domain", "model", "rcm", "rcm_version", "ensemble", "experiment"]]
subset["rcm_version"] = subset["rcm"] + "_" + subset["rcm_version"]
subset["available"] = "X"
```

```python
all_possibilities = pd.MultiIndex.from_product([
        subset["domain"].unique(),
        subset["model"].unique(),
        subset["experiment"].unique(),
        subset["ensemble"].unique(),
        subset["rcm_version"].unique()],
    names=["domain", "model", "experiment", "ensemble", "rcm_version"]
).to_frame(index=False)

available = all_possibilities[["domain", "model", "ensemble", "experiment", "rcm_version"]].drop_duplicates().merge(
    subset[["domain", "model", "ensemble", "experiment", "rcm_version", "available"]],
    how="outer").drop_duplicates().pivot(
    index=["model", "experiment", "ensemble", "rcm_version"],
    columns="domain",
    values="available").dropna(how="all").fillna("")
```
